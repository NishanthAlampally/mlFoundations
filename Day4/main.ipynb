```python
def batch_gradient_descent(Input, Z, lr):
    Theta = np.random.uniform(-1, 1, Input.shape[1])
    for _ in range(100000):
        h_x_matrix = Input @ Theta
        Theta = Theta - lr * (Input.T @ (h_x_matrix - Z))
    return Theta
```

### Stochastic Gradient Descent

```python
def stochastic_gradient_descent(Input, Z, lr):
    Theta = np.random.uniform(-1, 1, Input.shape[1])
    for _ in range(10000):
        for i in range(Input.shape[0]):
            h_x_matrix = Input[i] @ Theta
            Theta = Theta - lr * (Input[i] * (h_x_matrix - Z[i]))
    return Theta
```

```python
X = np.array([random.uniform(0, 20) for _ in range(15)])
Y = np.array([random.uniform(0, 20) for _ in range(15)])
Input = np.column_stack((np.ones(15), X, Y))
rng = np.random.default_rng(seed=42)
noise = rng.normal(loc=0.0, scale=0.5, size=15)
Z = 2*X + 3*Y + 5 + noise
```

```python
Theta = np.linalg.inv(Input.T @ Input) @ Input.T @ Z # Closed form
```

```python
# Batch Gradient Descent with log likelihood
Theta = np.random.uniform(-1, 1, Input.shape[1])
lr = 0.001
sigma = 4
for _ in range(100000):
    Theta = Theta + lr * (Input.T @ (Z - Input @ Theta)) / (sigma**2)
```
